```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=data, x="LUNG_CANCER")

plt.title("Lung Cancer Distribution")
plt.xlabel("Lung Cancer")
plt.ylabel("Number of People")
plt.show()
```

```python
sns.histplot(data=data, x="AGE", bins=15, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(data=data, x="LUNG_CANCER", y="AGE")

plt.title("Age vs Lung Cancer")
plt.xlabel("Lung Cancer")
plt.ylabel("Age")
plt.show()
```

```python
sns.countplot(data=data, x="GENDER", hue="LUNG_CANCER")

plt.title("Gender vs Lung Cancer")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```
